# Problem 87 - Prime Power Triples

The smallest number expressible as the sum of a prime square, prime cube, and prime fourth power is $28$. In fact, there are exactly four numbers below fifty that can be expressed in such a way:
$$\begin{align}
28 = 2^2 + 2^3 + 2^4\\
33 = 3^2 + 2^3 + 2^4\\
49 = 5^2 + 2^3 + 2^4\\
47 = 2^2 + 3^3 + 2^4
\end{align}$$
How many numbers below fifty million can be expressed as the sum of a prime square, prime cube, and prime fourth power?
## Solution.

```python
from math import sqrt
```

```python
N = 50_000_000
```

```python
primes = [True] * (int(sqrt(N)) + 1)
primes[0], primes[1] = False, False

for p in range(2, len(primes), 1):
    if primes[p] == True:
        for k in range(p**2, len(primes), p):
            primes[k] = False

primes = [p for p, _ in enumerate(primes) if primes[p] ==True]
```

```python
ans = 0
seen = set()
primes2 = [p for p in primes if p**2 < N]
primes3 = [p for p in primes if p**3 < N]
primes4 = [p for p in primes if p**4 < N]

for p in primes2:
    for q in primes3:
        for r in primes4:
            if p**2 + q**3 + r**4 < N and p**2 + q**3 + r**4 not in seen:
                ans += 1
                seen.add( p**2 + q**3 + r**4)

print(ans)
```
