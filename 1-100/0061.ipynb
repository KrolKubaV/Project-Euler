# Problem 61 - Cyclical Figurate Numbers

Triangle, square, pentagonal, hexagonal, heptagonal, and octagonal numbers are all figurate (polygonal) numbers and are generated by the following formulae:

| Type | Formula | Sequence |
| --- | --- | --- |
| Triangle | $P_{3,n}=n(n+1)/2$ | $1, 3, 6, 10, 15, \dots$ |
| Square | $P_{4,n}=n^2$ | $1, 4, 9, 16, 25, \dots$ |
| Pentagonal | $P_{5,n}=n(3n-1)/2$ | $1, 5, 12, 22, 35, \dots$ |
| Hexagonal | $P_{6,n}=n(2n-1)$ | $1, 6, 15, 28, 45, \dots$ |
| Heptagonal | $P_{7,n}=n(5n-3)/2$ | $1, 7, 18, 34, 55, \dots$ |
| Octagonal | $P_{8,n}=n(3n-2)$ | $1, 8, 21, 40, 65, \dots$ |

The ordered set of three $4$-digit numbers: $8128$, $2882$, $8281$, has three interesting properties.

1. The set is cyclic, in that the last two digits of each number is the first two digits of the next number (including the last number with the first).

2. Each polygonal type: triangle ($P_{3,127}=8128$), square ($P_{4,91}=8281$), and pentagonal ($P_{5,44}=2882$), is represented by a different number in the set.

3. This is the only set of $4$-digit numbers with this property.

Find the sum of the only ordered set of six cyclic $4$-digit numbers for which each polygonal type: triangle, square, pentagonal, hexagonal, heptagonal, and octagonal, is represented by a different number in the set.

## Solution.

```python
from collections import defaultdict

fig = defaultdict(list)

def generate(key, formula):
    n = 1
    while True:
        x = formula(n)
        if x >= 10000:
            break
        if x >= 1000:
            fig[key].append(str(x))
        n += 1

generate('t3', lambda n: n*(n+1)//2)
generate('t4', lambda n: n*n)
generate('t5', lambda n: n*(3*n-1)//2)
generate('t6', lambda n: n*(2*n-1))
generate('t7', lambda n: n*(5*n-3)//2)
generate('t8', lambda n: n*(3*n-2))


def matches(a, b):
    return a[2:] == b[:2]
```

```python
types = list(fig.keys())


def search(path, remaining):
    # completed cycle
    if not remaining:
        if matches(path[-1], path[0]):
            print(path, sum(map(int, path)))
        return

    last = path[-1]

    for t in remaining:
        for x in fig[t]:
            if matches(last, x):
                search(path + [x], remaining - {t})



for t in types:
    for x in fig[t]:
        search([x], set(types) - {t})
```

```python
def find_cycle(V, E):

    path = []
    types = set()

    def dfs(node, start):

        node_type = node[:2]

        # Cannot use the same figurate type twice
        if node_type in types:
            return None

        path.append(node)
        types.add(node_type)

        # We have chosen 6 numbers
        if len(path) == 6:

            # Check if last connects back to first
            if start in E[node]:
                return path + [start]

            # backtrack
            path.pop()
            types.remove(node_type)
            return None


        for neighbor in E[node]:

            result = dfs(neighbor, start)

            if result:
                return result


        # backtrack
        path.pop()
        types.remove(node_type)

        return None


    for start in V:
        result = dfs(start, start)

        if result:
            return result

    return None
```

```python
find_cycle(V, E)
```
