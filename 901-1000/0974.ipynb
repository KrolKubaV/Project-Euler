# Problem 974 - Very Odd Numbers
A <i>very odd number</i> is a number which contains only odd digits and is divisible by $105$.
Further each odd digit occurs an odd number of times.

Define $\Theta (n)$ be the $n$th very odd number, then $\Theta (1) = 1117935$ and $\Theta(10^3) = 11137955115$.

Find $\Theta(10^{16})$.
## Solution. 

```python
from functools import cache
```

```python
@cache
def ways(rem, mod7, mod3, mask):
    digits = [1, 3, 5, 7, 9]
    
    if rem == 0:
        return int(mod7 == 0 and mod3 == 0 and mask == (2**5-1))

    total = 0
    for i, d in enumerate(digits):
        if rem == 1 and d != 5:
            continue

        n7 = (10*mod7 + d) % 7
        n3 = (10*mod3 + d) % 3
        nm = mask ^ (1 << i)
        total += ways(rem-1, n7, n3, nm)

    return total
```

```python
def number_of_digits(N):
    digits = [1, 3, 5, 7, 9]
    L = 5
    total = 0
    
    while total < N:
        total += sum(ways(L-1, d%7, d%3, (1<<i)) for i, d in enumerate(digits))
        L += 2
        

    return L - 2
```

```python
number_of_digits(1), number_of_digits(10**3)
```

```python
def Theta(N):
    digits = [1, 3, 5, 7, 9]
    L = number_of_digits(N)
    N = N - sum(sum(ways(K-1, d%7, d%3, (1<<i)) for i, d in enumerate(digits)) for K in range(7, L, 2))

    mod7 = 0
    mod3 = 0
    mask = 0
    ans = []

    for pos in range(L):
        rem = L - pos

        for i, d in enumerate(digits):
            if rem == 1 and d != 5:
                continue

            n7 = (mod7*10 + d) % 7
            n3 = (mod3 + d) % 3
            nmask = mask ^ (1 << i)
            
            cnt = ways(rem-1, n7, n3, nmask)

            if cnt < N:
                N -= cnt
            else:
                ans.append(str(d))
                mod7, mod3, mask = n7, n3, nmask
                break

    return int("".join(ans))
```
