# Problem 987 - Straight Eight
In Poker a <b>straight</b> is exactly five cards of sequential rank NOT all of the same suit.
In this problem an ace can rank either high as in A-K-Q-J-10 or low as in 5-4-3-2-A, but cannot simultaneously rank both high and low, so Q-K-A-2-3 is not allowed.

There are $10200$ ways of choosing a straight from a normal $52$ card deck.
There are $31832952$ ways of choosing two disjoint straights from a single $52$ card deck.


Find the number of ways of choosing eight disjoint straights from a single $52$ card deck.
In this the order of choosing the straights does not matter.

## Solution.

```python
from math import comb, factorial
```

```python
def prod(arr):
    ans = 1

    for a in arr:
        ans *= a
    
    return ans
```

```python
def number_of_straights(deck):
    ans = 0
    
    for start in deck:
        if start > 9:
            break

    
        ans += (prod([len(deck[i%13]) for i in range(start, start+5)]) - 4)

    return ans
```

```python
def generate_tuples(N):
    n = 13
    x = [None]*n
    results = []

    fixed = {10:0, 11:0, 12:0}
    for i,v in fixed.items():
        x[i] = v

    def valid(i):
        if x[i] is None:
            return True

        # forward constraint
        for k in range(1,5):
            j = (i+k) % n
            if x[j] is not None and x[j] > 4 - x[i]:
                return False

        # backward constraint
        for k in range(1,5):
            j = (i-k) % n
            if x[j] is not None and x[i] > 4 - x[j]:
                return False

        return True

    def backtrack(i, current_sum):
        if current_sum > N:
            return

        if i == n:
            if current_sum == N:
                results.append(tuple(x))
            return

        if x[i] is not None:
            if valid(i):
                backtrack(i+1, current_sum + x[i])
            return

        for v in range(5):
            x[i] = v
            if valid(i):
                backtrack(i+1, current_sum + v)
            x[i] = None

    backtrack(0, 0)
    return results
```
