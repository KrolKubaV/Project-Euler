# Problem 991 - Fruit Salad

Find the sum of $x + y + z$ for all triples of positive integers $x, y, z$ which satisfy

$$
\frac{x}{y + z} +
\frac{y}{z + x} +
\frac{z}{x + y} = 4
$$

and

$$
x + y + z \le 10^7.
$$

## Solution.

```python
from tqdm import tqdm
```

```python
LIMIT = 10**7

# smallest prime factor sieve
spf = list(range(LIMIT))
for i in range(2, int(LIMIT**0.5) + 1):
    if spf[i] == i:
        for j in range(i*i, LIMIT, i):
            if spf[j] == j:
                spf[j] = i

def get_divisors_of_square(n):
    factors = {}
    while n > 1:
        p = spf[n]
        factors[p] = factors.get(p, 0) + 1
        n //= p

    divisors = [1]
    for p, e in factors.items():
        new = []
        for d in divisors:
            val = 1
            for _ in range(2*e + 1):
                new.append(d * val)
                val *= p
        divisors = new
    return divisors
```

```python
total = 0

for b in tqdm(range(2, LIMIT)):
    b2 = b * b
    for c in get_divisors_of_square(b):
        a = 4*b - b2 // c
        if 4*c > b and c-a > 0 and b-c+a > 0:
            s = 5*b - b2 // c
            if s <= LIMIT:
                a = 4*b - b**2/c
                total += s

                if a/b + b/c != 4:
                    print(a,b,c, a/b+b/c)

print(total)
```
