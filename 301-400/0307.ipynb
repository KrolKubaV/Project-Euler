# Problem 307 - Chip Defects
$k$ defects are randomly distributed amongst $n$ integrated-circuit chips produced by a factory (any number of defects may be found on a chip and each defect is independent of the other defects).

Let $p(k, n)$ represent the probability that there is a chip with at least $3$ defects.

For instance $p(3,7) \approx 0.0204081633$.

Find $p(20\,000, 1\,000\,000)$ and give your answer rounded to $10$ decimal places in the form 0.abcdefghij.


## Solution.
We find the formula for $p(k,n)$ as 
$$p(k, n)=1 - \frac{1}{n^k}\sum_{k_1+2k_2=k} \binom{n}{k_1, k_2, n-k_1-k_2} \cdot \frac{k!}{2^{k_2}}.$$
Note that if $a_k=\sum_{k_1+2k_2=k} \binom{n}{k_1, k_2, n-k_1-k_2} \cdot \frac{1}{2^{k_2}}$ and $f(x)=\left(1+x+\frac{1}{2}x^2\right)^n$ then it is easy to see that
$$\sum_{k_1+2k_2=k} \binom{n}{k_1, k_2, n-k_1-k_2} \cdot \frac{k!}{2^{k_2}}=k!\cdot a_k = f^{(k)}(0).$$
We then find a recursion for $f^{(k)}(x)$ and put $x=0$ to obtain a recursion for $f^{(k)}(0)$.

```python
from math import comb, factorial
from functools import cache
```

```python
def p(k, n):
    s = 0

    for k1 in range(k+1):
        k2 = (k-k1)/2
        if k2 == int(k2):
            k2 = int(k2)
            s += (comb(n, int(k2)) * comb(n-k2, k1))/(2**(k2)) 

    return round(1 - s * factorial(k)/n**k, 10)

p(3, 7)
```

```python
def a(k, n):
    if k == 0:
        return 1
    
    prev, cur = 1, n              

    for i in range(2, k + 1):
        prev, cur = cur, (n - i + 1) * cur + (i - 1) * (2*n - i + 2) // 2 * prev

    return cur

def p(k, n):
    return round(1 - a(k, n) / n**k, 10)
```

```python
p(20_000, 1_000_000)
```

```python
a(20_000, 1_000_000)
```
